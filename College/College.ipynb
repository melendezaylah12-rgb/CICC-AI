```python
mean_tuition_revenue_per_student = df["tuition_revenue_per_student"].mean()
mean_tuition_revenue_per_student
```

```python
boulder = df[df["school_name"].str.contains("Boulder")]
boulder = df[df["school_name"] == "University of Colorado Boulder"]
df_boulder = boulder["tuition_revenue_per_student"].item()
df_boulder
```

```python
Stanford = df[df["school_name"] == "Stanford University"]
df_Stanford = Stanford["tuition_revenue_per_student"].item()
df_Stanford
```

```python
z_boulder = (df_boulder - mean_tuition_revenue_per_student) / one_std_tuition_revenue_per_student
z_boulder
```

```python
z_Stanford = (df_Stanford - mean_tuition_revenue_per_student) / one_std_tuition_revenue_per_student
z_Stanford
```

# Conclusion
Based on the Z-Scores that were calculated we see that the z-score for Stanford is more higher with a score of 1.93 while Boulder has a z-score of 1.12 so we can understand that for the net tuition revenue for full time students Stanford has the highest z-score.
