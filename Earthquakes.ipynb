```python
df = pd.json_normalize(data["features"])
N = len(df)
print(f"Target population size (N): {N}")
df.head(100)
```

```python
conv_sample = df.head(100).copy()
```

```python
srs_sample = df.sample(n=100, random_state=42).copy()

def get_stats(dataframe, label):
    mean_mag = dataframe["properties.mag"].mean()
    times = pd.to_datetime(dataframe["properties.time"], unit="ms")
    time_span = times.max() - times.min()
    print(f"--- {label} ---")
    print(f"Mean Magnitude: {mean_mag:.4f}")
    print(f"Time Span: {time_span} (from {times.min()} to {times.max()})\n")
```

```python
get_stats(df, "True Population (N)")
get_stats(conv_sample, "Convenience Sample (First 100)")
get_stats(srs_sample, "Simple Random Sample (n=100)")
```

The USGS API returns earthquakes sorted chronologically in descending order (most recent events first) by default. Taking the first 100 rows captures only a narrow window of time—typically just the final few hours or days of the 30-day period—rather than the entire month. Because seismic activity is heavily clustered in both time and space (such as an earthquake swarm or a major mainshock followed by dozens of aftershocks in a single region), a non-random slice of the most recent events introduces selection bias. If an aftershock sequence or regional cluster occurred right before the end of the query window, those events will dominate the convenience sample and skew the sample mean away from the true population mean.

## Region

```python
import numpy as np
import pandas as pd
import requests

url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
params = {
    "format": "geojson",
    "eventtype": "earthquake",
    "minmagnitude": 2.0,
    "starttime": "2026-10-01",
    "endtime": "2026-10-31",
    "limit": 20000
}

response = requests.get(url, params=params)
data = response.json()
df = pd.json_normalize(data["features"])
```

```python
srs_sample = df.sample(n=100, random_state=42).copy()
```

```python
df["region"] = df["properties.place"].fillna("Unknown").apply(
    lambda x: x.split(",")[-1].strip() if "," in x else x.strip()
)

na_regions = {
    "CA", "California", "AK", "Alaska", "HI", "Hawaii", "NV", "Nevada",
    "WA", "Washington", "OR", "Oregon", "UT", "Utah", "ID", "Idaho",
    "MT", "Montana", "WY", "Wyoming", "TX", "Texas", "OK", "Oklahoma",
    "NM", "New Mexico", "AZ", "Arizona", "PR", "Puerto Rico",
    "Mexico", "Canada", "Dominican Republic", "Guatemala", "El Salvador",
    "Nicaragua", "Costa Rica", "Panama", "U.S. Virgin Islands"
}

asia_regions = {
    "Japan", "Indonesia", "Philippines", "Taiwan", "China", "Russia",
    "India", "Nepal", "Iran", "Afghanistan", "Pakistan", "Tajikistan",
    "Kyrgyzstan", "Uzbekistan", "Myanmar", "Thailand", "Timor-Leste",
    "South Korea", "Kuril Islands", "Bonin Islands", "Izu Islands", "Ryukyu Islands"
}

europe_regions = {
    "Greece", "Italy", "Turkey", "Iceland", "Spain", "Portugal",
    "Romania", "Albania", "Croatia", "Bosnia and Herzegovina",
    "France", "Switzerland", "Norway", "Azores", "Svalbard and Jan Mayen"
}

def map_stratum(reg):
    if reg in na_regions:
        return "North America"
    elif reg in asia_regions:
        return "Asia"
    elif reg in europe_regions:
        return "Europe"
    else:
        return "South America / Africa / Oceania"

df["stratum"] = df["region"].apply(map_stratum)
print("Population Stratum Counts:\n", df["stratum"].value_counts(), "\n")
# this would give us the Stratum count of each country
```

```python
n_total = 200
strata_props = df["stratum"].value_counts(normalize=True)

raw_counts = strata_props * n_total
sample_counts = np.floor(raw_counts).astype(int)
remainder = n_total - sample_counts.sum()
for stratum in (raw_counts - sample_counts).nlargest(remainder).index:
    sample_counts[stratum] += 1

stratified_sample = pd.concat([
    df[df["stratum"] == stratum].sample(n=count, random_state=42)
    for stratum, count in sample_counts.items()
])

print("Stratified Sample Stratum Counts:\n", stratified_sample["stratum"].value_counts(), "\n")
```

```python
srs_200 = df.sample(n=200, random_state=42)

var_mean_srs_200 = srs_200["properties.mag"].var(ddof=1) / len(srs_200)
var_mean_srs_100 = srs_sample["properties.mag"].var(ddof=1) / len(srs_sample)

var_mean_strat = 0
for stratum, count in sample_counts.items():
    W_h = strata_props[stratum]
    s_h_sq = stratified_sample[stratified_sample["stratum"] == stratum]["properties.mag"].var(ddof=1)
    var_mean_strat += (W_h ** 2) * (s_h_sq / count)

print(f"Stratified Sample Mean Magnitude: {stratified_sample['properties.mag'].mean():.4f}")
print(f"SRS (n=200) Mean Magnitude:       {srs_200['properties.mag'].mean():.4f}\n")
print(f"Variance of Mean (Stratified, n=200): {var_mean_strat:.6f}")
print(f"Variance of Mean (SRS, n=200):        {var_mean_srs_200:.6f}")
print(f"Variance of Mean (SRS, n=100):        {var_mean_srs_100:.6f}")
# This gives us the varience of the mean for each of the countries.
```

The variance of the stratified sample's mean magnitude is **lower** than the variance of the simple random sample's mean magnitude

* **Removal of Between-Stratum Variance:** Earthquake magnitudes differ systematically across geographic regions due to local tectonic settings and seismic sensor coverage (for example, dense regional networks in North America capture many small magnitude ~2.0 quakes, while remote oceanic or subduction zones record a higher proportion of larger events). By grouping the population into regional strata and drawing proportionally from each, stratified sampling eliminates the variation *between* regions from the sampling error—leaving only the *within-stratum* variance.
* **Guaranteed Proportional Representation:** A Simple Random Sample (SRS) can randomly over- or under-sample high- or low-magnitude regions in any single draw, leading to greater sample-to-sample variability. Proportional stratification locks in each region's exact share of the population, producing a more precise estimate with a smaller variance of the sample mean-SRS due to doubling the sample size.
